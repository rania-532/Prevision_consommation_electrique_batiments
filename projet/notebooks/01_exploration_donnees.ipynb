# 01 - Exploration des données

## 1. Importation des bibliothèques et chargement des datasets

Cette étape consiste à importer les bibliothèques nécessaires et à charger les trois fichiers CSV :
- electricity.csv : données de consommation électrique.
- metadata.csv : informations sur les bâtiments.
- weather.csv : données météorologiques.

```python
import sys
import pandas as pd

print("Python:", sys.executable)
print("Pandas:", pd.__version__)
```

```python

# Importation des bibliothèques
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Configuration de l'affichage
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 20)
pd.set_option("display.float_format", "{:.2f}".format)

# Définition du chemin des données
DATA_DIR = Path("../data/raw")

# Chargement des datasets
electricity = pd.read_csv(DATA_DIR / "electricity.csv")
metadata = pd.read_csv(DATA_DIR / "metadata.csv")
weather = pd.read_csv(DATA_DIR / "weather.csv")

print("Les trois datasets ont été chargés avec succès !")
```

```python
print("Electricity :", electricity.shape)
print("Metadata    :", metadata.shape)
print("Weather     :", weather.shape)
```

```python

# Exploration générale des trois datasets

datasets = {
    "ELECTRICITY": electricity,
    "METADATA": metadata,
    "WEATHER": weather
}

for name, df in datasets.items():
    print("\n" + "=" * 70)
    print(f"DATASET : {name}")
    print("=" * 70)

    print("\nDimensions :", df.shape)

    print("\nColonnes :")
    print(df.columns.tolist()[:20])

    print("\nTypes des données :")
    print(df.dtypes.value_counts())

    print("\nPremières lignes :")
    display(df.head())
```
